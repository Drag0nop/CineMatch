### Text Preprocessing

```python
# Cleaning the 'text' column by converting to lowercase, replacing non-word characters and newlines with spaces
df['text'] = df['text'].str.lower().replace(r'^\w\s', ' ').replace(r'\n', ' ', regex=True)
```

```python
df.tail(5)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

stemmer = PorterStemmer()

# Tokenize and stem the text
def tokenize(text):
    tokens = nltk.word_tokenize(text)
    stemming = [stemmer.stem(ch) for ch in tokens]
    return ''.join(stemming)
```

```python
# Applying the tokenize function to the 'text' column of the DataFrame
df['text'] = df['text'].apply(lambda x: tokenize(x))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tfidvec = TfidfVectorizer(analyzer='word', stop_words='english')
matrix = tfidvec.fit_transform(df['text'])
similarity = cosine_similarity(matrix)
```

```python
similarity[0]
```

```python
# Function to recommend songs based on the input song
df[df['song'] == 'Darkness']
```

```python
# Recommender Function
def recommend(song_name):
    # Finding the index of the song in the DataFrame
    idx = df[df['song'] == song_name].index[0]
    dist = sorted(list(enumerate(similarity[idx])), reverse=True, key=lambda x: x[1])

    songs = []

    # Getting the top 20 similar songs based on cosine similarity
    for i in dist[1:21]:
        # Appending the song names to the list of recommended songs
        songs.append(df.iloc[i[0]].song)
    return songs
```

```python
# Recommending songs similar to 'Darkness'
recommend('Darkness')
```
